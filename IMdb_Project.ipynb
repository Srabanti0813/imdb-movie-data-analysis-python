# Q3. How do we clean the duration column? (e.g., '175 min' → 175)

```python
df["duration"] = df["duration"].str.replace("min" , '')
df["duration"] = df["duration"].str.replace("," , '')

df["duration"] = df["duration"].astype(int) #Converting the datatype to integer
df.info()
```

# Q4. How do we clean the gross column and votes column?

```python
df["gross_income"] = df["gross_income"].str.replace("," , "")
df["gross_income"] = df["gross_income"].str.replace("$" , "")
df["gross_income"] = df["gross_income"].str.replace("M" , "")

df["gross_income"] = df["gross_income"].astype(float)
df.info()
```

```python
df["votes"] = df["votes"].str.replace("," , "")

df["votes"] = df["votes"].astype(int)

df.info()
```

# Q24. What is the best genre for a debut director? (Highest avg rating for 1st time directors)

```python

dir_counts = df.assign(directors_name=df["directors_name"].str.split(",")).explode("directors_name")["directors_name"].value_counts()
debut_directors = dir_counts[dir_counts == 1].index

df_debut = df.assign(
    directors_name=df["directors_name"].str.split(","),
    genre=df["genre"].str.split(",")
).explode("directors_name").explode("genre")

df_debut[df_debut["directors_name"].isin(debut_directors)].groupby("genre")["rating"].mean().sort_values(ascending=False).head(1)
```

# Q26. Which Director has the highest "Hit Rate"? (Percentage of movies rated > 7.5)

```python
x = df.assign(director=df["directors_name"].str.split(",")).explode("director")
x["director"] = x["director"].str.strip()
x["hit"] = x["rating"] > 7.5

y = x.groupby("director")["hit"].agg(["mean", "count"])
y[y["count"] >= 5]["mean"].mul(100).sort_values(ascending=False).head()
```

# Q27. Analyse the "Sequel Effect" — Do movies with numbers in titles earn more?

```python
df["has_number"] = df["name"].str.contains(r"\d")

df.groupby("has_number")["gross_income"].mean().sort_values(ascending=False)
```

# Q28. The Ultimate Business Decision: What movie should we fund?

```python

x = df.assign(genre=df["genre"].str.split(",")).explode("genre")
x["genre"] = x["genre"].str.strip()

y = x.groupby("genre").agg(
    avg_gross=("gross_income", "mean"),
    avg_rating=("rating", "mean"),
    total_movies=("name", "count")
)
y.sort_values(by="avg_gross", ascending=False).head()
```

#Q4. Which are the top 10 frequent actors? (Bar Chart)

```python
actors = df["stars_name"].str.split(",").explode().str.strip()

top10 = actors.value_counts().head(10)
```

```python
plt.figure(figsize = (12 , 6))

plt.bar(top10.index , top10.values , width = 0.7 , color = ["red" , "yellow" , "green" , "blue"])

plt.xlabel("Actors")
plt.ylabel("Numbers of movies")

plt.title("top 10 frequent actors")

plt.xticks(rotation = 45)

plt.show()
```

```python
#To understand it better

plt.figure(figsize = (12 , 6))

plt.barh(top10.index , top10.values , height = 0.5 , color = ["red" , "yellow" , "green" , "blue"])

plt.xlabel("Actors")
plt.ylabel("Numbers of movies")

plt.title("top 10 frequent actors")

plt.show()
```
